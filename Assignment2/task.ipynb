Write a Python program that calculates how much a project will cost.
Ask the user to enter:
Number of team members.
Number of months the project will take.
Monthly salary .
Total cloud cost for the whole project.
Then calculate:
Salary cost = team members × months × monthly salary
Subtotal = salary cost + cloud cost
VAT = subtotal × 0.14
Total = subtotal + VAT

Print the , using f-strings with :,.2f to show amounts like 12,500.00.


```python
num_of_team_members = int(input("Enter the number of team members: "))
num_of_months = float(input("Enter the number of months the project will take: "))
# انا قاصد احط القيم بالعشري عشان احسب التكلفة بدقة
Monthly_salary = float(input("Enter the monthly salary of each team member: "))
cloud_cost = float(input("Enter the Total cloud cost for the whole project: "))
```

```python
Salary_cost = num_of_team_members * num_of_months * Monthly_salary
subtotal = Salary_cost + cloud_cost
VAT = subtotal * 0.14
total_ = subtotal + VAT                                         
```

```python
print(f"Salary Cost: {Salary_cost:,.2f}")
print(f"Subtotal: {subtotal:,.2f}")
print(f"VAT: {VAT:,.2f}")
print(f"Total Cost: {total_:,.2f}")
```
