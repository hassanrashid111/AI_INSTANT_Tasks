### 1. Maps Application: Google Maps

* **<font color="blue">Data Received</font>:** Real-time GPS location of smartphones, historical traffic data, movement speed of other drivers, and user report.

* **<font color="blue">AI Task</font>:** using machine learning to analyze traffic patterns and find the fastest route.

* **<font color="blue">Result Produced</font>:** Estimated Time of Arrival and the best recommended driving route.

## Question 2: Business Problems Classification

Classify each business problem as best suited to:
- **AI** (Machine Learning, Computer Vision, NLP)
- **Traditional Software** (Rule-based programming, fixed logic)
- **Human** (Empathy, negotiations, personal care)

| # | Business Problem | Best Suited To | Justification |
|---|---|---|---|
| **1** | **Predicting next month’s sales from historical data.** | **AI (Machine Learning)** | Past sales data has complex seasonal patterns and trends that machine learning models learn and forecast accurately. |
| **2** | **Calculating invoice totals using fixed tax and discount rules.** | **Traditional Software** | This task follows fixed mathematical rules, so standard code calculates exact numbers with 100% accuracy without needing AI. |
| **3** | **Resolving a sensitive conflict between two employees.** | **Human** | Resolving employee conflicts requires empathy, active listening, and emotional understanding that machines do not possess. |
| **4** | **Recommending products based on customers’ browsing and purchase history.** | **AI (Machine Learning)** | Customer preferences involve large, dynamic datasets where AI recommendation systems excel at finding patterns and personalizing suggestions. |
