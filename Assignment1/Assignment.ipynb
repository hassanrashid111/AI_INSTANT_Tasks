Analyze real-world AI applications (**Maps**, **Banking**, **Shopping**, and **Gmail Spam Filter**) using the AI System Analysis Canvas:
1. What data does the system receive?

### 1. Maps Application: Google Maps

* **<font color="blue">Data Received</font>:** Real-time GPS location of smartphones, historical traffic data, speed limits, movement speed of other drivers, and user reports (accidents, road work, closures).

* **<font color="blue">AI Task</font>:** Predictive traffic analysis and route optimization (using machine learning to analyze traffic patterns and find the fastest route).

* **<font color="blue">Result Produced</font>:** Estimated Time of Arrival (ETA) and the best recommended driving route.

* **<font color="blue">How Output is Used</font>:** Displays turn-by-turn navigation on the driver's screen and suggests faster alternative routes when heavy traffic is ahead.

* **<font color="blue">Potential Risks</font>:**
  1. **Routing Errors:** Sudden road closures or map mistakes can guide drivers into dangerous roads or heavy traffic jams.
  2. **Location Privacy:** Continuously tracking GPS location collects sensitive personal information about where users live, work, and visit.

### 2. Banking Application: Fraud Detection System

* **<font color="blue">Data Received</font>:** Transaction amount, store name, purchase time, location/country, user device details, and previous customer spending habits.

* **<font color="blue">AI Task</font>:** Anomaly detection and classification (spotting transactions that look different from normal customer behavior).

* **<font color="blue">Result Produced</font>:** A risk score classifying the transaction as safe, suspicious, or high risk.

* **<font color="blue">How Output is Used</font>:** Normal transactions are approved instantly, while suspicious transactions trigger an SMS/OTP verification or are blocked immediately to protect the user.

* **<font color="blue">Potential Risks</font>:**
  1. **False Positives:** The system may mistakenly block a genuine purchase, locking a customer's card during an emergency or while traveling.
  2. **Undetected Fraud (False Negatives):** Smart scammers may bypass the AI, causing financial loss to the customer.

### 3. Shopping Application: Amazon Product Recommendations

* **<font color="blue">Data Received</font>:** Browsing history, items added to cart or wishlist, past purchases, search keywords, product reviews, and shopping behavior of similar customers.

* **<font color="blue">AI Task</font>:** Recommendation engine using collaborative filtering (learning customer preferences to predict what items they might want to buy).

* **<font color="blue">Result Produced</font>:** A personalized list of recommended products (e.g., *"Frequently bought together"*, *"Customers also bought"*).

* **<font color="blue">How Output is Used</font>:** Shown on the homepage, product pages, and checkout screen to encourage customers to discover and buy more products.

* **<font color="blue">Potential Risks</font>:**
  1. **Privacy Concerns:** Tracking user browsing habits and personal interests can feel invasive to customers.
  2. **Irrelevant or Annoying Suggestions:** The system might keep recommending products the user already bought in physical stores or clicked by mistake.

### 4. Email Application: Gmail Spam Filter

* **<font color="blue">Data Received</font>:** Email text and subject lines, sender domain reputation, embedded links, attachments, and user actions (marking emails as spam or not spam).

* **<font color="blue">AI Task</font>:** Natural Language Processing (NLP) and text classification to detect spam words, scams, and phishing attempts.

* **<font color="blue">Result Produced</font>:** A probability score indicating whether an incoming email is safe, spam, or phishing.

* **<font color="blue">How Output is Used</font>:** Automatically routes safe emails to the Primary Inbox and sends unwanted or malicious emails to the Spam folder.

* **<font color="blue">Potential Risks</font>:**
  1. **False Positives:** Important legitimate emails (such as job offers, university letters, or bank alerts) might go to Spam and get missed.
  2. **Phishing Scams (False Negatives):** New and smart phishing attacks might slip into the Inbox, risking user account security.
